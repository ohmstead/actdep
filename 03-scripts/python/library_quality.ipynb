This notebook will visually summarize the sublibrary data quality immediately after sequencing. I want to plot metrics such as read depth, counts, genes, reads mapped, etc. This will rely heavily on data from the Parse pipeline.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(rc={'font.family': 'avenir', 'font.size': 12})
plt.rcParams['figure.dpi'] = 600
```

```python
summary_meta1 = pd.read_csv('/Volumes/jack/seq/analysis_May2024/sublibrary1/0_all-sample/agg_samp_ana_summary_sublibrary1.csv')
summary_meta1.set_index('statistic', drop=True, inplace=True)
summary_meta1.index.name = None
summary_meta1 = summary_meta1.T
summary_meta1['sublibrary'] = 1

summary_meta2 = pd.read_csv('/Volumes/jack/seq/analysis_May2024/sublibrary2/0_all-sample/agg_samp_ana_summary_sublibrary2.csv')
summary_meta2.set_index('statistic', drop=True, inplace=True)
summary_meta2.index.name = None
summary_meta2 = summary_meta2.T
summary_meta2['sublibrary'] = 2

# combine sublibraries, drop sample average rows
summ = pd.concat([summary_meta1, summary_meta2])
summ = summ.drop('all-sample')

summ['sample'] = summ.index
summ['condition'] = summ['sample'].apply(lambda x: x[:-1])

summ
```

```python
p = sns.swarmplot(summ, 
                  x='condition',
                  y='number_of_cells', 
                  hue='sublibrary', dodge=True, size=15, palette=sns.color_palette())
p.set_ylim(0, 1500)
p.set_ylabel('')
p.set_xlabel('')
p.set_xticklabels(labels= ['dSE', 'd30m', 'd6h', 'KA'], fontsize=15)
p.set_title('Nuclei per sample', fontsize=20)

plt.figure()
p4 = sns.stripplot(summ, 
                  x='condition', 
                  y='sequencing_saturation',
                  hue='sublibrary', 
                  dodge=True, size=15, palette=sns.color_palette(), legend=False)
p4.set_ylim(0, 1)
p4.set_xlabel('')
p4.set_ylabel('')
p4.set_xticklabels(p.get_xticklabels(), fontsize=15)
p4.set_title('Seq saturation', fontsize=20)
```

```python
# fig, ax = plt.subplots(1, 4, figsize=(16,9))
plt.figure()
p1 = sns.stripplot(summ, 
                  x='condition', 
                  y='mean_reads_per_cell',
                  hue='sublibrary', 
                  dodge=True, size=15, palette=sns.color_palette(), legend=False)
p1.set_ylim(0, 120000)
p1.set_xlabel('')
p1.set_ylabel('')
p1.set_xticklabels(labels= ['1h', '6h', 'SE'], fontsize=15)
p1.set_title('Reads per cell', fontsize=20)

plt.figure()
p2 = sns.stripplot(summ, 
                  x='condition', 
                  y='mm39_median_tscp_per_cell',
                  hue='sublibrary', 
                  dodge=True, size=15, palette=sns.color_palette(), legend=False)
p2.set_ylim(0, 25000)
p2.set_xlabel('')
p2.set_ylabel('')
p2.set_xticklabels(labels= ['1h', '6h', 'SE'], fontsize=15)
p2.set_title('Median transcripts per cell', fontsize=20)

plt.figure()
p3 = sns.stripplot(summ, 
                  x='condition', 
                  y='mm39_median_genes_per_cell',
                  hue='sublibrary', 
                  dodge=True, size=15, palette=sns.color_palette(), legend=False)
p3.set_ylim(0, 6000)
p3.set_xlabel('')
p3.set_ylabel('')
p3.set_xticklabels(labels= ['1h', '6h', 'SE'], fontsize=15)
p3.set_title('Median genes per cell', fontsize=20)

plt.figure()
p4 = sns.stripplot(summ, 
                  x='condition', 
                  y='sequencing_saturation',
                  hue='sublibrary', 
                  dodge=True, size=15, palette=sns.color_palette(), legend=False)
p4.set_ylim(0, 1)
p4.set_xlabel('')
p4.set_ylabel('')
p4.set_xticklabels(p.get_xticklabels(), fontsize=15)
p4.set_title('Seq saturation', fontsize=20)
```

Now load in information about pre-filtered cells

```python
cells = pd.read_csv('/Volumes/jack/seq/analysis/combined/0_all-sample/DGE_filtered/cell_metadata.csv')
cells['condition'] = cells['sample'].apply(lambda x: x[:2])
cells.set_index('bc_wells', drop=True, inplace=True)
cells.index.name = None
cells['sublibrary'] = cells.index.to_series().apply(lambda x: x[-1:]).astype('int')
cells
```

```python
cells.columns
```

```python
f, ax = plt.subplots(3, 1, sharex=True, figsize=(16,9), dpi=600, )
a = sns.violinplot(cells[cells['mread_count'] < 125000],
                   ax=ax[0],
                   x='sample',
                   y='mread_count',
                   hue='sublibrary',
                   split=True,
                   inner='quart',
                   palette=sns.color_palette(),
                   cut=1,
                #    log_scale=True,
                   )
ax[0].set_ylim(0)
ax[0].set_ylabel('Reads', fontsize=20)
ax[0].set_xlabel('')
ax[0].set_yticks(ax[0].get_yticks()[1:-1])
# ax[0].legend(loc='upper left')


sns.violinplot(cells[cells['tscp_count'] < 40000],
                   ax=ax[1],
                   x='sample',
                   y='tscp_count',
                   hue='sublibrary',
                   split=True,
                   inner='quart',
                   palette=sns.color_palette(),
                #    log_scale=True,
                   legend=False
                   )
ax[1].set_ylim(0)
ax[1].set_ylabel('Transcripts', fontsize=20)
ax[1].set_xlabel('')
ax[1].set_yticks(ax[1].get_yticks()[1:-1])

ax[2] = sns.violinplot(cells[cells['gene_count'] < 12000],
                   ax=ax[2],
                   x='sample',
                   y='gene_count',
                   hue='sublibrary',
                   split=True,
                   inner='quart',
                   palette=sns.color_palette(),
                   log_scale=False,
                   legend=False,
                   )
ax[2].set_ylim(0, 15000)
ax[2].set_ylabel('Genes', fontsize=20)
ax[2].set_xlabel('')
ax[2].set_yticks(( 5000, 10000, ))
ax[2].set_xticklabels(['1h1', '1h2', '6h1', '6h2', 'SE1', 'SE2', 'SE3'], fontsize=20)


plt.subplots_adjust(hspace=0.01)
```
