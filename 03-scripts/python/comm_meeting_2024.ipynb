```python
iegs = [
'Fosb',
'Npas4',
'Fos',
'Junb',
'Nr4a1',
'Egr4',
'Arc',
'Egr2',
'Egr1',
'Dusp1',
'Dusp5',
'Egr3',
'Btg2',
'Ppp1r15a',
]
```

```python
order = ['dSE', 'd30m', 'd6h', 'KA']
sc.pl.dotplot(adata, iegs, groupby='condition', swap_axes=False, categories_order=order)
sc.pl.matrixplot(adata, iegs, groupby='condition', swap_axes=True, categories_order=order)
sc.pl.stacked_violin(adata, iegs, groupby='condition', swap_axes=True, categories_order=order)
# plot a heatmap with the order above
sc.pl.heatmap(adata, iegs, groupby='condition', swap_axes=True, categories_order=order)
```

```python
# filter out KA from adata
order = ['dSE', 'd30m', 'd6h', 'KA']
adata_noKA = adata[adata.obs['condition'] != 'KA']
sc.pl.heatmap(adata_noKA, iegs, groupby='condition', swap_axes=True)
```

```python
for gene in iegs:
    sc.pl.umap(adata, color=[gene], show=False)
    ax = plt.gca()
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.set_xlabel('')
    ax.set_ylabel('')
    ax.set_title(ax.get_title(), fontsize=30, fontdict={'fontstyle': 'italic'})

    # Remove color bar if present
    colorbar = ax.collections[0].colorbar
    if colorbar is not None:
        colorbar.remove()
```
